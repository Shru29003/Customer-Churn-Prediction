```python
import pandas as pd
import numpy as np

df = pd.read_csv("Customer_Churn_Prediction_8000_Records.csv")
```

```python
print (df [ "TenureMonths"].describe())
```

```python
print(df[["TenureMonths", "Tenure_Group"]].head(10))

print(df["Tenure_Group"].value_counts())
```

```python
print(df["MonthlyCharges"].describe())
```

```python
low_threshold = df["MonthlyCharges"].quantile(0.25)
high_threshold = df["MonthlyCharges"].quantile(0.75)

print("Low threshold:", low_threshold)
print("High threshold:", high_threshold)
```

```python
df["Spending_category"] = np.where(
    df["MonthlyChargers"] <= low_threshold,
    "Low Spending",
    np.where(
        df["MonthlyCharger"] <= high_threshold,
        "Medium Spending ",
        "High Spending "
    ))
```

```python
print(df[["MonthlyCharges", "Spending_Category"]].head(10))

print(df["Spending_Category"].value_counts())
```

```python
print(df["Tenure_Group"].value_counts())
```

```python
missing_count = df.isnull().sum()
print ("Missing Values Count:")
print(missing_count)
```

```python
missing_percentage = (
    df.isnull().sum() / len(df)
) * 100
print ("Missing Values Percentage:")
print(missing_percentage)
```

```python
df["OnlineBackup"] = df["OnlineBackup"].fillna(
    df["OnlineBackup"].mode()[0]
)
```
